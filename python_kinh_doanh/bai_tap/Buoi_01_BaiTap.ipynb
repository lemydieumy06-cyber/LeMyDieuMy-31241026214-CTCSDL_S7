### A2. Tính giá bán sau chiết khấu và thuế VAT
**Tình huống:** Cửa hàng bán một chiếc tai nghe giá niêm yết 850.000 đ, giảm 15%, sau đó cộng VAT 8%. Khách mua 3 chiếc.

**Bước 1 – Khai báo biến.** Tên biến viết thường, nối bằng dấu `_`, đặt tên có nghĩa.

```python
gia_niem_yet = 850_000      # đồng (dấu _ giúp dễ đọc số lớn)
ty_le_giam = 0.15           # 15%
vat = 0.08                  # 8%
so_luong = 3
print(type(gia_niem_yet), type(ty_le_giam), type(so_luong))
```

**Bước 2 – Tính toán** với các toán tử `+ - * /`.

```python
gia_sau_giam = gia_niem_yet * (1 - ty_le_giam)
gia_sau_vat = gia_sau_giam * (1 + vat)
thanh_tien = gia_sau_vat * so_luong
print(gia_sau_giam, gia_sau_vat, thanh_tien)
```

**Bước 3 – Định dạng kết quả.** f-string `f"...{bien:,.0f}..."` chèn giá trị vào chuỗi;
`:,.0f` = dấu phẩy ngăn cách hàng nghìn, 0 chữ số thập phân. Thay `,` bằng `.` để đúng kiểu Việt Nam.

```python
def vnd(x):
    return f"{x:,.0f}".replace(",", ".") + " đ"

print("Giá sau giảm :", vnd(gia_sau_giam))
print("Giá sau VAT  :", vnd(gia_sau_vat))
print("Thành tiền   :", vnd(thanh_tien))
```

### A3. Phép chia lấy nguyên `//`, chia lấy dư `%` và lũy thừa `**`
**Tình huống:** Kho có 1.250 chai nước, mỗi thùng đóng 24 chai. Được bao nhiêu thùng, dư bao nhiêu chai?

```python
tong_chai = 1250
chai_moi_thung = 24
so_thung = tong_chai // chai_moi_thung
du = tong_chai % chai_moi_thung
print(f"Đóng được {so_thung} thùng, còn dư {du} chai")
```

**Lãi kép:** Gửi 100 triệu đồng, lãi suất 6%/năm, sau 5 năm được bao nhiêu? Công thức: `A = P × (1 + r)^n`.

```python
P, r, n = 100_000_000, 0.06, 5
A = P * (1 + r) ** n
print("Số tiền sau", n, "năm:", vnd(A))
print("Tiền lãi:", vnd(A - P))
```

## Phần B. Bài tập tự làm

Mỗi bài viết trong ô code ngay bên dưới đề. In kết quả rõ ràng, có đơn vị.

### Bài B1. Hóa đơn bán hàng

Khách mua 2 áo sơ mi (350.000 đ/chiếc) và 1 quần tây (520.000 đ). Cửa hàng giảm 10% trên tổng tiền hàng,
sau đó tính VAT 8%. In ra: tổng tiền hàng, tiền giảm, tiền VAT, số tiền khách phải trả (định dạng `1.234.567 đ`).

```python
# Bài B1: Hóa đơn bán hàng

gia_ao = 350_000
so_ao = 2
gia_quan = 520_000
ty_le_giam = 0.10
vat = 0.08

tong_tien_hang = gia_ao * so_ao + gia_quan
tien_giam = tong_tien_hang * ty_le_giam
tien_sau_giam = tong_tien_hang - tien_giam
tien_vat = tien_sau_giam * vat
khach_phai_tra = tien_sau_giam + tien_vat

def dinh_dang_vnd(so_tien):
    return f"{so_tien:,.0f}".replace(",", ".") + " đ"

print("Tổng tiền hàng :", dinh_dang_vnd(tong_tien_hang))
print("Tiền giảm giá  :", dinh_dang_vnd(tien_giam))
print("Tiền VAT       :", dinh_dang_vnd(tien_vat))
print("Khách phải trả :", dinh_dang_vnd(khach_phai_tra))
```

### Bài B2. Điểm hòa vốn

Một quán cà phê có định phí 45 triệu đồng/tháng, giá bán trung bình 35.000 đ/ly, biến phí 14.000 đ/ly.
Tính số ly cần bán mỗi tháng để hòa vốn (làm tròn lên) và số ly mỗi ngày (tháng 30 ngày).
Công thức: `Q = Định phí / (Giá bán − Biến phí)`.

<details><summary>💡 Gợi ý</summary>

Dùng `math.ceil()` để làm tròn lên (cần `import math`).

</details>

```python
import math

dinh_phi = 45_000_000       # đồng/tháng
gia_ban = 35_000             # đồng/ly
bien_phi = 14_000            # đồng/ly

so_ly_hoa_von = math.ceil(dinh_phi / (gia_ban - bien_phi))
so_ly_moi_ngay = math.ceil(so_ly_hoa_von / 30)

print("Số ly cần bán mỗi tháng để hòa vốn (được làm tròn):", so_ly_hoa_von, "ly")
print("Số ly cần bán mỗi ngày (30 ngày):", so_ly_moi_ngay, "ly")
```

### Bài B3. Trả góp hàng tháng

Khách mua laptop 24.000.000 đ trả góp 12 tháng, lãi suất 1,2%/tháng. Tính số tiền trả mỗi tháng theo công thức
niên kim: `PMT = P × r × (1 + r)^n / ((1 + r)^n − 1)`, tổng số tiền trả và tổng tiền lãi.

```python
gia_laptop = 24_000_000
lai_suat_thang = 0.012
so_thang = 12

# Công thức trả góp đều (niên kim)
pmt = gia_laptop * lai_suat_thang * (1 + lai_suat_thang) ** so_thang / ((1 + lai_suat_thang) ** so_thang - 1)
tong_tien_tra = pmt * so_thang
tong_tien_lai = tong_tien_tra - gia_laptop

def dinh_dang_vnd(so_tien):
    return f"{so_tien:,.0f}".replace(",", ".") + " đ"

print("Tiền trả mỗi tháng:", dinh_dang_vnd(pmt))
print("Tổng số tiền trả  :", dinh_dang_vnd(tong_tien_tra))
print("Tổng tiền lãi     :", dinh_dang_vnd(tong_tien_lai))
```

### Bài B4. Chuẩn hóa mã sản phẩm

Cho chuỗi `ma = "  sp-dt-00045 "`. Hãy: (a) bỏ khoảng trắng và viết hoa → `SP-DT-00045`;
(b) tách lấy mã nhóm hàng `DT`; (c) lấy số thứ tự dưới dạng số nguyên `45`;
(d) tạo mã mới cho sản phẩm kế tiếp `SP-DT-00046` (giữ đủ 5 chữ số).

<details><summary>💡 Gợi ý</summary>

`f"{so:05d}"` in số nguyên đủ 5 chữ số, thêm số 0 phía trước.

</details>

```python
ma = "  sp-dt-00045 "
ma_chuan = ma.strip().upper()
cac_phan = ma_chuan.split("-")
ma_nhom_hang = cac_phan[1]
so_thu_tu = int(cac_phan[2])
ma_tiep_theo = f"{cac_phan[0]}-{ma_nhom_hang}-{so_thu_tu + 1:05d}"

print("Mã chuẩn hóa :", ma_chuan)
print("Nhóm hàng    :", ma_nhom_hang)
print("Số thứ tự    :", so_thu_tu)
print("Mã kế tiếp   :", ma_tiep_theo)
```

### Bài B5. Tăng trưởng doanh thu

Doanh thu quý 1 là 1,85 tỷ đồng, quý 2 là 2,12 tỷ đồng. Tính tỷ lệ tăng trưởng (%) và in dạng `Tăng trưởng: 14.6%`.
Nếu duy trì tốc độ này, doanh thu quý 4 dự kiến bao nhiêu?

```python
# Bài B5: Tăng trưởng doanh thu

doanh_thu_q1 = 1.85  # tỷ đồng
doanh_thu_q2 = 2.12  # tỷ đồng

ty_le_tang_truong = (doanh_thu_q2 - doanh_thu_q1) / doanh_thu_q1

# Nếu duy trì cùng tốc độ tăng trưởng, dự kiến quý 3 và quý 4
doanh_thu_q3 = doanh_thu_q2 * (1 + ty_le_tang_truong)
doanh_thu_q4 = doanh_thu_q3 * (1 + ty_le_tang_truong)

print(f"Tăng trưởng: {ty_le_tang_truong:.1%}")
print(f"Doanh thu quý 4 dự kiến: {doanh_thu_q4:.2f} tỷ đồng")
```

### Bài C1. So sánh hai phương án gửi tiết kiệm

Phương án 1: lãi 6,2%/năm, nhập lãi hằng năm. Phương án 2: lãi 6,0%/năm nhưng nhập lãi hằng tháng
(lãi tháng = 6,0%/12). Gửi 200 triệu đồng trong 3 năm. Phương án nào có lợi hơn và hơn bao nhiêu?

```python
tien_gui = 200_000_000
so_nam = 3

# Phương án 1: lãi 6,2%/năm, nhập lãi hằng năm
lai_suat_1 = 0.062
tien_pa1 = tien_gui * (1 + lai_suat_1) ** so_nam

# Phương án 2: lãi danh nghĩa 6%/năm, nhập lãi hằng tháng
lai_suat_thang_2 = 0.06 / 12
so_thang = so_nam * 12
tien_pa2 = tien_gui * (1 + lai_suat_thang_2) ** so_thang

chenh_lech = abs(tien_pa1 - tien_pa2)

def dinh_dang_vnd(so_tien):
    return f"{so_tien:,.0f}".replace(",", ".") + " đ"

print("Số tiền phương án 1:", dinh_dang_vnd(tien_pa1))
print("Số tiền phương án 2:", dinh_dang_vnd(tien_pa2))
if tien_pa1 > tien_pa2:
    print("Vậy thì Phương án 1 có lợi hơn:", dinh_dang_vnd(chenh_lech))
else:
    print("Phương án 2 có lợi hơn:", dinh_dang_vnd(chenh_lech))
```

### Bài C3. Quy tắc 72

Quy tắc 72 ước lượng số năm để vốn tăng gấp đôi: `72 / lãi suất (%)`. Với lãi suất 4%, 6%, 8%, 12%,
hãy so sánh kết quả ước lượng với kết quả chính xác `ln(2) / ln(1 + r)`. (Có thể viết lặp lại 4 lần;
buổi sau ta sẽ dùng vòng lặp.)

```python
import math

# Quy tắc 72 dùng lãi suất tính theo phần trăm.
print("Lãi suất | Quy tắc 72 (năm) | Công thức chính xác (năm)")
for lai_suat_phan_tram in [4, 6, 8, 12]:
    r = lai_suat_phan_tram / 100
    uoc_tinh_72 = 72 / lai_suat_phan_tram
    so_nam_chinh_xac = math.log(2) / math.log(1 + r)
    print(f"{lai_suat_phan_tram:>7}% | {uoc_tinh_72:>16.2f} | {so_nam_chinh_xac:>23.2f}")
```
